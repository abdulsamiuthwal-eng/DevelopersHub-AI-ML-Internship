```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
iris =sns.load_dataset("iris")
print(iris.head())
```

#📦 Step 1: Importing Libraries and Loading Dataset
We import pandas, seaborn, and matplotlib for data loading and visualization.
The Iris dataset is loaded directly using seaborn's built-in datasets.

```python
print(iris.shape)
```

#📐 Step 2: Dataset Shape
The dataset contains 150 rows and 5 columns.
4 columns are features and 1 column is the target (species).

```python
print (iris.info())
```

#📊 Step 3: Scatter Plot
A scatter plot to visualize the relationship between sepal length and petal length,
colored by species to show class separation.

```python
print(iris.describe())
```

#📊 Step 4: Statistical Summary
describe() shows count, mean, std, min and max of all numeric columns.
Petal length has highest variation (std = 1.76) and sepal width is most consistent (std = 0.43).

```python
sns.scatterplot(x="sepal_length", y="petal_length", hue="species", data=iris)

plt.title("Scatter Plot of Iris Dataset")

plt.show()
```

#🌸 Step 5: Scatter Plot
Scatter plot shows relationship between sepal length and petal length.
Setosa species is clearly separated while Versicolor and Virginica slightly overlap.

```python
iris.hist(figsize=(10,8))
plt.show()
```

#📈 Step 6: Histogram
Histograms show the distribution of each feature.
Petal length and petal width show bimodal distribution indicating two groups in data.

```python
sns.boxplot(data=iris)
plt.title("BOXPLOT OF IRIS DATASET")
plt.show()
```

#📦 Step 7: Box Plot
Box plot helps identify outliers in each feature.
Sepal width has 2 outliers visible and petal length has the widest range among all features.
